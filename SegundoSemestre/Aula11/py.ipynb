#Estrutura de dados em Json


```python
import json 

data = {
    'nome': 'joao',
    'idade' : 20,
    'cidade' : 'sao paulo'
}
data
```

#Escrever dados 


```python
with open ('exemplo_json.json','w') as json_file:
    json.dump(data, json_file) 
```

#Ler dados

```python
with open ('exemplo_json.json', 'r') as json_file:
    df_json = json.load(json_file)
```

```python
def buscar_info(nome, data):
    if nome in data["nome"]:
        index = data["nome"].index(nome)
        return f"Nome: {nome}, Idade: {data['idade'][index]}, Profissão: {data['profissao'][index]}"
    else:
        return "Nome não encontrado."

print(buscar_info("Gabriel", informacoes))
```
